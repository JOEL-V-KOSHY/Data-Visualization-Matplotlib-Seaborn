```python

# ============================================================
# PROJECT 02: DATA VISUALIZATION WITH MATPLOTLIB & SEABORN
# InternCircle - Data Science & AI
# Dataset: Seaborn Tips Dataset
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

from IPython.display import display

print("Libraries imported successfully!")


# ============================================================
# 2. SET VISUALIZATION THEME
# ============================================================

sns.set_theme(
    style="whitegrid",
    palette="deep",
    font_scale=1.1
)

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100

print("Visualization theme configured!")


# ============================================================
# 3. LOAD DATASET
# ============================================================

tips = sns.load_dataset("tips")

print("\nDATASET LOADED SUCCESSFULLY")
print("=" * 50)

print("First Five Records:")
display(tips.head())

print("\nDataset Shape:", tips.shape)
print("Number of Rows:", tips.shape[0])
print("Number of Columns:", tips.shape[1])

print("\nColumn Names:")
print(tips.columns.tolist())


# ============================================================
# 4. DATA EXPLORATION
# ============================================================

print("\nDATASET INFORMATION")
print("=" * 50)

tips.info()

print("\nSTATISTICAL SUMMARY")
display(tips.describe())

print("\nMISSING VALUES")
print(tips.isnull().sum())

print("\nDUPLICATE RECORDS")
print(tips.duplicated().sum())


# ============================================================
# 5. DATA CLEANING
# ============================================================

tips = tips.drop_duplicates().copy()

tips = tips.dropna().copy()

print("\nDATA CLEANING COMPLETED")
print("Updated Dataset Shape:", tips.shape)
print("Remaining Missing Values:", tips.isnull().sum().sum())


# ============================================================
# 6. CREATE OUTPUT DIRECTORY
# ============================================================

output_folder = "project_02_outputs"

os.makedirs(output_folder, exist_ok=True)

print("\nOutput directory created:", output_folder)


# ============================================================
# 7. HISTOGRAM - DISTRIBUTION OF TOTAL BILLS
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True,
    color="#3498DB",
    edgecolor="white"
)

plt.title(
    "Distribution of Restaurant Bills",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Total Bill (USD)")
plt.ylabel("Number of Meals")

plt.tight_layout()

plt.savefig(
    f"{output_folder}/chart1_histogram.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 8. BAR CHART - AVERAGE TIPS BY DAY
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="day",
    estimator="mean",
    errorbar=None,
    palette="Set2",
    legend=False
)

plt.title(
    "Average Tips Across Different Days",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Day")
plt.ylabel("Average Tip (USD)")

plt.tight_layout()

plt.savefig(
    f"{output_folder}/chart2_barplot.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 9. BOX PLOT - TIP DISTRIBUTION BY DAY
# ============================================================

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    hue="day",
    palette="pastel",
    legend=False
)

plt.title(
    "Distribution of Tips Across Different Days",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Day")
plt.ylabel("Tip Amount (USD)")

plt.tight_layout()

plt.savefig(
    f"{output_folder}/chart3_boxplot.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. SCATTER PLOT - TOTAL BILL VS TIP
# ============================================================

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time",
    palette="Set1",
    s=80,
    alpha=0.8
)

plt.title(
    "Relationship Between Total Bill and Tip",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Total Bill (USD)")
plt.ylabel("Tip Amount (USD)")

plt.legend(title="Meal Time")

plt.tight_layout()

plt.savefig(
    f"{output_folder}/chart4_scatterplot.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 11. CORRELATION HEATMAP
# ============================================================

correlation = tips.select_dtypes(
    include="number"
).corr()

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues",
    linewidths=0.5,
    fmt=".2f",
    square=True
)

plt.title(
    "Correlation Between Numerical Variables",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    f"{output_folder}/chart5_heatmap.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 12. CUSTOM COLOR PALETTE VISUALIZATION
# ============================================================

custom_palette = ["#3498DB", "#2ECC71", "#F39C12", "#9B59B6"]

plt.figure(figsize=(9, 5))

sns.countplot(
    data=tips,
    x="day",
    hue="day",
    palette=custom_palette,
    legend=False
)

plt.title(
    "Number of Restaurant Visits by Day",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Day")
plt.ylabel("Number of Visits")

plt.tight_layout()

plt.savefig(
    f"{output_folder}/chart6_custom_palette.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 13. COMBINED 2x2 VISUALIZATION DASHBOARD
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# Histogram
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True,
    color="#3498DB",
    ax=axes[0, 0]
)

axes[0, 0].set_title("Distribution of Restaurant Bills")
axes[0, 0].set_xlabel("Total Bill (USD)")
axes[0, 0].set_ylabel("Number of Meals")


# Bar Chart
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="day",
    estimator="mean",
    errorbar=None,
    palette="Set2",
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set_title("Average Tips by Day")
axes[0, 1].set_xlabel("Day")
axes[0, 1].set_ylabel("Average Tip (USD)")


# Scatter Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time",
    palette="Set1",
    ax=axes[1, 0]
)

axes[1, 0].set_title("Total Bill vs Tip")
axes[1, 0].set_xlabel("Total Bill (USD)")
axes[1, 0].set_ylabel("Tip (USD)")


# Heatmap
sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues",
    fmt=".2f",
    ax=axes[1, 1]
)

axes[1, 1].set_title("Correlation Heatmap")


fig.suptitle(
    "Restaurant Tips - Data Visualization Dashboard",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.savefig(
    f"{output_folder}/tips_dashboard.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 14. AUTOMATIC DATA ANALYSIS AND FINDINGS
# ============================================================

print("\n")
print("=" * 65)
print("PROJECT FINDINGS AND DATA INSIGHTS")
print("=" * 65)

# Average bill
average_bill = tips["total_bill"].mean()

# Average tip
average_tip = tips["tip"].mean()

# Average tips by day
daily_tips = tips.groupby(
    "day", observed=True
)["tip"].mean()

# Bill and tip correlation
bill_tip_corr = tips["total_bill"].corr(tips["tip"])

# Strongest correlation excluding self-correlations
corr_pairs = correlation.where(
    np.triu(np.ones(correlation.shape), k=1).astype(bool)
).stack()

strongest_pair = corr_pairs.abs().idxmax()
strongest_value = corr_pairs.loc[strongest_pair]

print(f"\n1. Average Restaurant Bill: ${average_bill:.2f}")

print(f"\n2. Average Tip: ${average_tip:.2f}")

print("\n3. Average Tips by Day:")
print(daily_tips.round(2))

print(
    f"\n4. Bill and Tip Correlation: {bill_tip_corr:.2f}"
)

print(
    "\n5. Strongest Numerical Correlation:",
    f"{strongest_pair[0]} and {strongest_pair[1]}",
    f"({strongest_value:.2f})"
)


# ============================================================
# 15. WRITTEN CHART TAKEAWAYS
# ============================================================

print("\n")
print("=" * 65)
print("CHART TAKEAWAYS")
print("=" * 65)

print(
    "\n1. HISTOGRAM:",
    f"The average restaurant bill is ${average_bill:.2f}.",
    "The histogram shows the distribution of restaurant bills."
)

print(
    "\n2. BAR CHART:",
    f"{daily_tips.idxmax()} has the highest average tip,",
    f"at ${daily_tips.max():.2f}."
)

print(
    "\n3. BOX PLOT:",
    "The box plot compares the median, spread and outliers",
    "of tip amounts across different days."
)

print(
    "\n4. SCATTER PLOT:",
    f"The correlation between total bill and tip is {bill_tip_corr:.2f}.",
    "This indicates a positive or negative linear association."
)

print(
    "\n5. HEATMAP:",
    f"The strongest numerical relationship is between",
    f"{strongest_pair[0]} and {strongest_pair[1]},",
    f"with a correlation of {strongest_value:.2f}."
)


# ============================================================
# 16. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("=" * 65)
print("PROJECT SUMMARY")
print("=" * 65)

print("Project: Data Visualization with Matplotlib & Seaborn")
print("Dataset: Seaborn Tips Dataset")
print("Total Records:", len(tips))
print("Total Features:", len(tips.columns))

print("\nVisualizations Created:")
print("1. Histogram")
print("2. Bar Chart")
print("3. Box Plot")
print("4. Scatter Plot")
print("5. Correlation Heatmap")
print("6. Custom Color Palette Chart")
print("7. Combined 2x2 Dashboard")

print("\nLibraries Used:")
print("Pandas, NumPy, Matplotlib, Seaborn")

print("\nProject Execution Completed Successfully!")


# ============================================================
# 17. DOWNLOAD ALL OUTPUT IMAGES AS ZIP
# ============================================================

import shutil

shutil.make_archive(
    "Project_02_Visualization_Charts",
    "zip",
    output_folder
)

print("\nAll visualization images saved successfully!")
print("ZIP file: Project_02_Visualization_Charts.zip")

# Google Colab download
try:
    from google.colab import files
    files.download("Project_02_Visualization_Charts.zip")
except ImportError:
    print("Download the ZIP file from the notebook file browser.")
```
